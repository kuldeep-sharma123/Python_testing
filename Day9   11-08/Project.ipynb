### Downloads the Domino's dataset from Kaggle.

```python
import pandas as pd
import kagglehub

path = kagglehub.dataset_download("avijitjana101/dominos-predictive-purchase-order-system")

print("Path to dataset files:", path)
```

### Lists all files in the downloaded dataset directory.


```python
import os
for root,dirs,files in os.walk(path):
    for file in files:
        print(file)
        
```

### Loads the Excel dataset into a Pandas DataFrame.

```python
import pandas as pd
import os

csv_file = os.path.join(path, "Pizza_Sale.xlsx")

df = pd.read_excel(csv_file)
df
```

### Removes duplicate records based on the `order_id` column.

```python
df1=df.drop_duplicates(subset="order_id")
df1
```

### Checks for missing values in each column.

```python
df1.isnull().sum()
```

### Removes rows containing missing values.

```python
df1=df1.dropna()
df1
```

### Displays the data types of all columns.

```python
df1.dtypes
```

### Converts the `order_date` column to datetime format.

```python
df1["order_date"] = pd.to_datetime(df1["order_date"])
df1
```

### Filters records where the quantity is greater than 1.

```python
df2=df1[df1["quantity"] > 1]
df2
```

### Calculates the total quantity sold for each pizza and sorts the results in descending order.

```python
df2.groupby("pizza_name_id")["quantity"].sum().sort_values(ascending=False)
```

### Sorts the dataset by unit price in descending order.

```python
df2.sort_values(by="unit_price", ascending=False)
```

### Selects and rearranges the required columns for further analysis.

```python
df2 = df2[[
    "pizza_id",
    "pizza_name_id",
    "quantity",
    "order_date",
    "order_time",
    "unit_price",
    "total_price",
    "pizza_size",
    "pizza_category",
    "pizza_ingredients",
    "pizza_name"
]]

df2
```

### Exports the cleaned dataset to a CSV file.

```python
df2.to_csv("Dominos_cleaned_Data.csv")
```

### Imports the required libraries for data analysis and visualization.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### Creates a bar chart showing the total quantity sold by each pizza category.

```python
category = df.groupby("pizza_category")["quantity"].sum()

plt.figure(figsize=(8,5))
colors=["limegreen","yellow"]
plt.bar(category.index, category.values,color=colors)

plt.title("Quantity by Category")
plt.xlabel("Category")
plt.ylabel("Quantity")

plt.show()
```

### Creates a pie chart showing the distribution of pizza sizes.

```python
size = df["pizza_size"].value_counts()

plt.figure(figsize=(6,6))
plt.pie(size.values,
        labels=size.index,
        autopct="%1.1f%%")

plt.title("Pizza Size Distribution")

plt.show()
```

### Creates a line chart showing the daily sales trend over time.

```python
df["order_date"] = pd.to_datetime(df["order_date"])

daily_sales = df.groupby("order_date")["total_price"].sum()

plt.figure(figsize=(12,5))
plt.plot(daily_sales.index,
         daily_sales.values,color="aqua",
         marker="o")

plt.title("Daily Sales Trend")
plt.xlabel("Date")
plt.ylabel("Sales")

plt.xticks(rotation=45)

plt.show()
```

### Creates a box plot to compare the distribution of total prices across different pizza sizes.

```python
plt.figure(figsize=(8,5))

sns.boxplot(data=df,
            x="pizza_size",
            y="total_price")

plt.title("Price by Pizza Size")

plt.show()
```

### Creates a heatmap to visualize the correlation between numerical features.

```python
corr = df.corr(numeric_only=True)

plt.figure(figsize=(6,5))

sns.heatmap(corr,
            annot=True,
            cmap="YlGnBu")

plt.show()
```

### Creates a pair plot to visualize relationships between quantity, unit price, and total price.

```python
sns.pairplot(df[["quantity","unit_price","total_price"]])

plt.show()
```
