```python
from matplotlib import pyplot as plt
product=["TV","Fridge","Ac"]
price=[12000,15000,20000]
plt.title("Sales Report")
plt.xlabel("product")
plt.ylabel("price")
plt.bar(product,price,color=["salmon","Aqua","Magenta"])
plt.show
```

```python
from matplotlib import pyplot as plt
product=["TV","Fridge","Ac"]
price=[12000,15000,20000]
plt.title("Sales Report")
plt.xlabel("product")
plt.ylabel("price")
plt.barh(product,price,color=["salmon","Aqua","Magenta"])
plt.show
```

```python
from matplotlib import pyplot as plt
product=["TV","Fridge","Ac"]
price=[12000,15000,20000]
plt.title("Sales Report")
plt.pie(
    price,
    labels=product,                      # labels, not label
    colors=["salmon", "aqua", "magenta"], # colors, not color
    autopct="%1.1f%%"                    # Optional: show percentages
)
plt.show
```

```python
from matplotlib import pyplot as plt
import pandas as pd
df=pd.read_csv("MegaMart.csv")
data=df.groupby("product_category")["quantity"].sum()


```

```python
cate=data.index
values=data.values
colors=["Blue","Red","Green","Yellow","Pink"]
plt.bar(cate,values,color=colors)
```
