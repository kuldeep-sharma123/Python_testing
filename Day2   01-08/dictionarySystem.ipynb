```python
Bank = {
    "canara": {"Rahul": 10000, "Mohan": 5000, "Abhinandan": 500},
    "SBI": {"Ram": 1000000},
    "HDFC": {"Kuldeep": 21000, "Binit": 20000}
}

print("Bank Details")
print(Bank)


Bank["canara"]["Shyam"] = 3000

print("After Adding Shyam")
print(Bank)

print("Highest Balance in Each Bank")

for bank, customers in Bank.items():
    highest_name = max(customers, key=customers.get)
    highest_balance = customers[highest_name]

    print(bank, " : ", highest_name, " : ", highest_balance)


bank_name = "canara"
customer = "Rahul"
amount = 5000

Bank[bank_name][customer] += amount

print("After Deposit")
print(customer, "Balance =", Bank[bank_name][customer])

amount = 2000

if Bank[bank_name][customer] >= amount:
    Bank[bank_name][customer] -= amount
    print("\nAfter Withdraw")
    print(customer, "Balance =", Bank[bank_name][customer])
else:
    print("Insufficient Balance")

print("Final Bank Data : ")
print(Bank)
```
