> #### Imported the required libraries and downloaded the dataset using KaggleHub.

```python
import pandas as pd
import kagglehub

path = kagglehub.dataset_download("mdsazzatsardar/amazonsalesreport")

print(path)
```

#### Used `os.walk()` to traverse the dataset folder and display all available files and subdirectories.

```python
import os
for root,dirs,files in os.walk(path):
    for file in files:
        print(file)
```

#### Loaded the Amazon Sales Report CSV file into a Pandas DataFrame using `pd.read_csv()` for data cleaning, analysis, and visualization.

```python
import pandas as pd
import os

csv_file = os.path.join(path, "Amazon Sale Report.csv")

df = pd.read_csv(csv_file)
df

```

#### Display the dimensions (rows and columns) of the dataset using `df.shape`.

```python
df.shape
```

#### Display the first 10 rows of the dataset using `df.head(10)` to quickly preview the data structure and verify that it has been loaded correctly.

```python
df.head(10)
```

#### Generate statistical summary of the dataset using `df.describe()` to understand the distribution, central tendency, and spread of numerical columns.

```python
df.describe()
```

#### Display the dataset information using `df.info()` to examine column names, data types, non-null values, and memory usage before performing data cleaning and analysis.

```python
df.info()
```

#### Calculate the percentage of missing values in each column to identify incomplete data and determine the appropriate data cleaning strategy.

```python

(df.isnull().sum()/len(df))*100
```

#### Fill missing numerical values with the column mean.

```python
numeric = df.select_dtypes(include="number").columns

for col in numeric:
    df[col] = df[col].fillna(df[col].mean())
df
```

#### Calculate the percentage of missing values in each column.

```python
(df.isnull().sum()/len(df))*100
```

#### Fill missing values using the next valid value with `bfill()`.

```python
df.bfill(inplace=True)
```

#### Verify the percentage of missing values after handling missing data.

```python
(df.isnull().sum()/len(df))*100
```

#### Rename column names by replacing spaces with underscores for better readability and easier access.

```python
df.columns = df.columns.str.replace(" ", "_")
df.info()
```

#### Sort the dataset by **ship-state** in ascending order and **Amount** in descending order to organize records efficiently.

```python
df=df.sort_values(by=["ship-state", "Amount"], ascending=[True, False])
df
```

#### Find the highest sales amount for each state using `groupby()` and `max()`.

```python
df.groupby("ship-state")["Amount"].max()
```

#### Find the lowest sales amount for each state using `groupby()` and `min()`.

```python
df.groupby("ship-state")["Amount"].min()
```

#### Save the cleaned dataset as a new CSV file using `to_csv()` for future analysis and sharing.

```python
df.to_csv(
    "Amazon_Cleaned.csv",
    index=False
)
```

#### Import Matplotlib and Pandas libraries for creating visualizations and analyzing the dataset.

```python
import matplotlib.pyplot as plt
import pandas as pd
```

#### Visualize the top 10 states by total sales using a bar chart.

```python
state_sales = df.groupby("ship-state")["Amount"].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,5))
colors = ["red", "orange", "yellow", "green", "blue", "indigo", "violet"]
plt.bar(state_sales.index, state_sales.values, color=colors)

plt.title("Top 10 States by Sales")
plt.xlabel("State")
plt.ylabel("Sales Amount")
plt.xticks(rotation=45)

plt.show()
```

#### Visualize the sales distribution across product categories using a pie chart.

```python
category = df.groupby("Category")["Amount"].sum().head(5)

plt.figure(figsize=(7,7))

plt.pie(
    category.values,
    labels=category.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Category Sales Distribution")

plt.show()
```

#### Visualize the distribution of sales amounts using a histogram.

```python
plt.figure(figsize=(8,5))

plt.hist(df["Amount"], bins=30, color="green", edgecolor="black")

plt.title("Amount Distribution")
plt.xlabel("Amount")
plt.ylabel("Frequency")

plt.show()
```

#### Visualize the relationship between quantity and sales amount using a scatter plot.

```python
plt.figure(figsize=(8,5))

plt.scatter(df["Qty"], df["Amount"],color="red")

plt.xlabel("Quantity")
plt.ylabel("Amount")

plt.title("Quantity vs Amount")

plt.show()
```

#### Visualize the monthly sales trend using a line chart.

```python
df["Date"] = pd.to_datetime(df["Date"])

monthly = df.groupby(df["Date"].dt.month)["Amount"].sum()

plt.figure(figsize=(8,5))

plt.plot(
    monthly.index,
    monthly.values,
    marker="o"
)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.grid(True)

plt.show()
```
