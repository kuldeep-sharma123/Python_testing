```python
import seaborn as sns
import matplotlib.pyplot as plt
tips=sns.load_dataset('tips')
 
print(tips)
 
sns.set(style="whitegrid")
 
plt.figure(figsize=(10,6))
ax=sns.barplot(x="day",y="total_bill",data=tips,estimator=sum,errorbar=None)   #setting ci=None disables the calculation and display of confidence intervals for estimated values in plots
 
ax.set_title('Total Bill amount per day')
ax.set_xlabel('Day')
ax.set_ylabel('Total Bill amount')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
 
# Load a sample dataset
df = sns.load_dataset('tips')
 
# 1. Scatter Plot
plt.figure(figsize=(7, 5))
sns.scatterplot(data=df, x='total_bill', y='tip')
plt.title('Scatter Plot of Total Bill vs Tip')
plt.show()
 
# 2. Line Plot
plt.figure(figsize=(7, 5))
sns.lineplot(data=df, x='size', y='total_bill')
plt.title('Line Plot of Group Size vs Total Bill')
plt.show()
 
# 3. Bar Plot
plt.figure(figsize=(7, 5))
sns.barplot(data=df, x='day', y='total_bill')
plt.title('Average Total Bill per Day')
plt.show()
 
# 4. Box Plot
plt.figure(figsize=(7, 5))
sns.boxplot(data=df, x='day', y='total_bill')
plt.title('Box Plot of Total Bill by Day')
plt.show()
 
# 5. Violin Plot
plt.figure(figsize=(7, 5))
sns.violinplot(data=df, x='day', y='total_bill')
plt.title('Violin Plot of Total Bill by Day')
plt.show()
 
# 6. Pair Plot
plt.figure(figsize=(7, 5))
sns.pairplot(df)
plt.title('Pair Plot of Tips Dataset')
plt.show()
 
# 7. Heatmap of Correlation Matrix
plt.figure(figsize=(7, 5))
corr = df.corr()
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Heatmap of Correlation Matrix')
plt.show()
 
# 8. Histogram with KDE
plt.figure(figsize=(7, 5))
sns.histplot(df['total_bill'], bins=20, kde=True)
plt.title('Histogram of Total Bill with KDE')
plt.show()
 
# 9. Facet Grid with Scatter Plot
g = sns.FacetGrid(df, col='sex', row='smoker')
g.map(sns.scatterplot, 'total_bill', 'tip')
plt.show()
```
