```python
class Book:
    def __init__(self,name,length):
        self.name=name
        self.length=length
    def description(self):
        print("The given book name is: ", self.name ,"And it contain :",self.length," pages")
b1 = Book("Harry and Potter","293")
b2=Book("Ramayan","3000")
b3=Book("sunderkand","400")
b1.description()
b2.description()
b3.description()
```
