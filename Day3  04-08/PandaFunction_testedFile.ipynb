```python
import pandas as pd
df=pd.read_csv("tested.csv")
df
```

Q.1.  Find the number of passengers in each passenger class (Pclass).

```python
df1=df.groupby("Pclass").size()
df1
```

Q.2. Find the average fare paid by passengers in each class.


```python
df2=df.groupby("Pclass")["Fare"].mean()
df2
```

Q.3. Find the maximum and minimum age in each class.

```python
df3=df.groupby("Pclass")["Age"].agg(["max", "min"])
df3
```

Q.4.  Find the number of male and female passengers.

```python
df4=df.groupby("Sex").size()
df4
```

Q.5. Find the average age of males and females.

```python
df5=df.groupby("Sex")["Age"].mean()
df5
```

Q.6.  Find the survival rate (mean of Survived) for each gender.

```python
df6=df.groupby("Sex")["Survived"].mean()
df6
```

Q.7. Find the number of passengers grouped by Pclass and Sex


```python
df7=df.groupby(["Pclass", "Sex"]).size()
df7
```

Q.8. Find the average age for each combination of Pclass and Survived.

```python
df8=df.groupby(["Pclass", "Survived"])["Age"].mean()
df8
```

Q.9. Find the minimum age for each passenger class and gender.

```python
df9=df.groupby(["Pclass", "Sex"])["Age"].min()
df9
```

Q.10. Find the sum, mean, and count of fares for each class.

```python
df10=df.groupby("Pclass")["Fare"].agg(["sum", "mean", "count"])
df10
```
