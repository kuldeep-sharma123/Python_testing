```python
import pandas as pd

data={'Student':['Alice','Riya','Aman','Hari'],
      'Class':['9A','6C','9A','6C'],
      'English':[84,78,70,81],
      'Science':[56,78,90,67]}
df=pd.DataFrame(data)

print("Original DataFrame:\n", df, "\n")

# 1. Sort by English marks (descending)
df_sorted = df.sort_values(by="English", ascending=False)
print("Sorted by English (descending):\n", df_sorted, "\n")

# 2. Drop the 'Class' column
df_dropped = df.drop(columns=['Class'])
print("After dropping 'Class' column:\n", df_dropped, "\n")

# 3. Rename column 'Science' → 'Science_Marks'
df_renamed = df.rename(columns={'Science': 'Science_Marks'})
print("After renaming column:\n", df_renamed, "\n")

# 4. Remove duplicates (if any)
df_no_duplicates = df.drop_duplicates()
print("After dropping duplicates:\n", df_no_duplicates, "\n")

# 5. Reset index after sorting
df_reset = df_sorted.reset_index(drop=True)
print("After resetting index:\n", df_reset)
```

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Aman", "Riya", "Karan"],
    "Marks": [80, 75, 90]
})

# Increase marks to 100 where marks are greater than 85
df.loc[df["Marks"] > 85, "Marks"] = 100

print(df)
```

```python
df.loc[df["Marks"] > 40, "Result"] = "Pass"
print(df)
```
