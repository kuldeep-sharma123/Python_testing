41. Create a Stock Status column (Out of Stock, Low, Medium, High).

```python
def stock_status(qty):
    if qty == 0:
        return "Out of Stock"
    elif qty <= 50:
        return "Low"
    elif qty <= 100:
        return "Medium"
    else:
        return "High"

df["Stock Status"] = df["qty"].apply(stock_status)
df
```

42. Create a 10% discounted price column.

```python
df["Discount_10%"]=df["unit_price"]*.1
df
```

43. Create a GST (18%) column.

```python
df["GST_18%"]=df["unit_price"]*.18
df
```

44. Calculate the final selling price after GST.

```python
df["Selling price"]=df["unit_price"]+df["GST_18%"]-df["Discount_10%"]
df
```

50. Create a summary report with totals, averages, highest/lowest priced products, categories, and suppliers.

```python
summary = {
    "Total Products": len(df),
    "Total Inventory Quantity": df["qty"].sum(),
    "Average Unit Price": df["unit_price"].mean(),
    "Highest Priced Product": df.loc[df["unit_price"].idxmax(), "item_name"],
    "Highest Price": df["unit_price"].max(),
    "Lowest Priced Product": df.loc[df["unit_price"].idxmin(), "item_name"],
    "Lowest Price": df["unit_price"].min(),
    "Total Categories": df["category"].nunique(),
    "Categories": df["category"].unique().tolist(),
    "Total Suppliers": df["supplier"].nunique(),
    "Suppliers": df["supplier"].dropna().unique().tolist()
}

for key, value in summary.items():
    print(f"{key}: {value}")
```
